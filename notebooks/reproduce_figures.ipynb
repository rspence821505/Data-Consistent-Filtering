# Reproducing the figures of *A Data-Consistent Approach to Ensemble Filtering*

This notebook is an interactive front end to **Level 1** of the repository: it redraws every figure of the paper from the saved numerical results, shows it next to the numbers the paper quotes, and checks it against the PNG used in the manuscript. It runs no experiment and takes well under a minute.

It calls exactly the code behind `python scripts/reproduce_figures.py`, so the two always agree. Set `RESULTS = "regenerated"` to draw Figures 1–5 from Lorenz-96 results you have recomputed with `scripts/run_all_experiments.py` (Level 2) instead of the published ones. The shallow-water experiment (Figures 6–8) is reproduced from its saved run outputs only, so those figures always use the published data.

| Figure | Content | Data |
|---|---|---|
| 1 | Spread–skill ratio, Lorenz-96, $N=10$ | `results/published/lorenz96/calibration_N10.npz` |
| 2 | Coverage, tail balance and persistence | `coverage_ranks_N10.npz`, `calibration_N10.npz` |
| 3 | Settled RMSE versus ensemble size | `ensemble_size_sweep.npz` |
| 4 | Hierarchical bias–variance composition | `hierarchical_bias_variance.npz` |
| 5 | Bias–variance over $(N,\kappa)$ | `n_kappa_bootstrap.npz` |
| 6 | Idealized-inlet configuration | `data/shallow_water/Ideal_Inlet.h5`, `shallow_water/scale_matching_regime_map_v3.csv` |
| 7 | Inlet error and spread–skill trajectories | `results/published/shallow_water/runs/*/result_qpca_endcf.json` |
| 8 | Inlet endpoint error and its composition | same run files |

```python
import sys
from pathlib import Path

# Make the repository importable whether the kernel starts in notebooks/ or the root.
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "src" / "figures").is_dir())
sys.path.insert(0, str(ROOT))

from IPython.display import Image, display

from src.figures import render
from src.figures.compare import compare_png, describe
from src.paths import GENERATED_FIGURES_DIR, PAPER_FIGURES_DIR, resolve_results_dir

RESULTS = "published"          # or "regenerated"
results_root = resolve_results_dir(RESULTS)
out_dir = GENERATED_FIGURES_DIR / f"notebook_{RESULTS}"


def show(number, width=900):
    """Redraw one paper figure, display it, print its numbers and compare it."""
    written, summary = render(number, results_root, out_dir)
    png = [p for p in written if p.suffix == ".png"][-1]
    display(Image(filename=str(png), width=width))
    print(summary)
    print("Comparison with the manuscript PNG:", describe(compare_png(png, PAPER_FIGURES_DIR / png.name)))


print("results:", results_root)
print("figures written to:", out_dir)
```
